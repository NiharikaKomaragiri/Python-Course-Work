```python
import pandas as pd
# Example: Product Prices
product_prices = pd.Series(
[2999, 15999, 52999, 4999, 1999],
index=["Wireless Earbuds", "Smartphone", "Laptop",
"Smartwatch", "Bluetooth Speaker"]
)
print(product_prices)
```

```python
print("Mean",product_prices.mean())
```

```python
print("Sum",product_prices.sum())
```

```python
print("Max",product_prices.max())
```

```python
print("Min",product_prices.min())
```

```python
print("Head(First 3 Elements):\n",product_prices.head(3))
```

```python
print("Tail(Last 2 Elements):\n",product_prices.tail(2))
```

```python
print("Apply(Adding 18% GST):\n",product_prices.apply(lambda x:f'{x * 1.18}'))
```

```python
print('Map(Formating as currency):\n',product_prices.map(lambda x:f'₹{x}') )
```

```python
print("Sort by values:\n",product_prices.sort_values())
```

```python
print("Sort by Index:\n",product_prices.sort_index())
```

```python
print(product_prices.sort_index(ascending=False))
```

```python
print("Value Counts:\n",product_prices.value_counts())
```

```python
# Example: Product Details
data = {
"Product": ["Wireless Earbuds", "Smartphone", "Laptop",
"Smartwatch", "Bluetooth Speaker"],
"Brand": ["SoundMax", "TechNova", "ByteCore", "TimeTrack",
"EchoBoom"],
"Price": [2999, 15999, 52999, 4999, 1999],
"Stock": [50, 30, 20, 40, 60],
"BestSeller": [True, False, True, True, False]
}
df = pd.DataFrame(data)
print(df)
```

```python
print("Shape of DataFrame:",df.shape)
```

```python
print("Columns in DataFrame:",df.columns)
```

```python
print("DataFrame INFO:")
print(df.info())
```
