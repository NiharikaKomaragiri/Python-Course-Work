```python
import matplotlib.pyplot as plt
import numpy as np
# Creating sample data
#Line Plot
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

```python
# Creating a figure and axis
fig, ax = plt.subplots()
ax.plot(x, y, label="Sine Wave", color='red', linestyle='--',
linewidth=2, marker='o')
```

```python
# Adding labels, title, legend, and grid
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_title("Basic Line Plot")
ax.legend()
ax.grid(True)
plt.show()
```

```python
plt.plot(x, y, color='red', linestyle='-', linewidth=2,marker='s', markersize=5, markerfacecolor='yellow')
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Line Plot with Attributes")
plt.grid(True)
plt.show()
```

```python
#Bar Graph
categories = ['A', 'B', 'C', 'D']
values = [10, 20, 15, 30]
plt.bar(categories, values, color='purple', edgecolor='black',linewidth=1.5, alpha=0.5)
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Bar Chart with Customization")
plt.show()
```

```python
#Histogram
#Used for distribution analysis.
data = np.random.randn(1000)
plt.hist(data, bins=30, color='cyan', edgecolor='black',alpha=0.8) #Defines the number of bins.
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram Example")
plt.show()
```

```python
#Scatter Plot
x = np.random.rand(50)
y = np.random.rand(50)
sizes = np.random.randint(10, 100, 50)
colors = np.random.rand(50)
```

```python
plt.scatter(x, y, s=sizes, c=colors, cmap='viridis',alpha=0.8, edgecolors='black')#cmap – Uses a colormap
plt.colorbar(label="color scale")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Scatter Plot with Custom Attributes")
plt.show()
```

```python
#Pie Chart
#autopct – Displays percentages.
#startangle – Rotates the chart.
#explode – Separates a slice for emphasis.
labels = ['Apples', 'Bananas', 'Cherries', 'Dates']
sizes = [30, 25, 35, 10]
colors = ['red', 'yellow', 'pink', 'purple']
plt.pie(sizes, labels=labels, colors=colors,autopct='%1.1f%%', startangle=140, explode=[0.1, 0, 0, 0])
plt.title("Pie Chart Example")
plt.show()
```

```python
#Box Plot
#Used for statistical distribution.

data = [np.random.randn(100), np.random.randn(100) * 2,
np.random.randn(100) * 1.5]
plt.boxplot(data, labels=['Group 1', 'Group 2', 'Group 3'],
patch_artist=True, boxprops=dict(facecolor="lightblue"))
plt.title("Box Plot Example")
plt.show()
#patch_artist – Fills the box with color.
#boxprops – Modifies box appearance.
```

```python
#Heatmap (Requires Seaborn)
#Used for correlation and density analysis.

import seaborn as sns
data = np.random.rand(10, 10)
sns.heatmap(data, annot=True, cmap="coolwarm")
plt.title("Heatmap Example")
plt.show()
#annot – Displays values in each cell.
#cmap – Changes the color scheme.
```

```python
#3D Scatter Plot
#Used for three-dimensional relationships.

from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
x = np.random.rand(50)
y = np.random.rand(50)
z = np.random.rand(50)
ax.scatter(x, y, z, c='red', marker='o')
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
ax.set_title("3D Scatter Plot")
plt.show()


#projection='3d' – Enables 3D plotting.
#ax.scatter() – Creates a 3D scatter plot.
```

```python
#Simple Area Chart

import matplotlib.pyplot as plt
import numpy as np
# Sample data
x = np.linspace(1, 10, 10)
y1 = np.array([3, 6, 8, 12, 15, 18, 20, 22, 25, 30])
y2 = np.array([2, 5, 7, 10, 13, 16, 18, 19, 22, 27])
# Creating the area chart
plt.fill_between(x, y1, color="blue", alpha=0.5,
label="Dataset 1")
plt.fill_between(x, y2, color="green", alpha=0.5,
label="Dataset 2")
# Adding labels, title, and legend
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Area Chart Example")
plt.legend()
plt.grid(True)
plt.show()
```
