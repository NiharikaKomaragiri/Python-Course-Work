```python
arr = np.array([10, 20, 30, 40, 50])
# Boolean indexing
bool_arr = arr > 25
print(bool_arr) # Returns [False, False, True, True, True]
# Filtering values
filtered_arr = arr[arr > 25]
print(filtered_arr) # Returns [30, 40, 50]
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
```

```python
# Matrix multiplication
print(np.dot(A, B))

# Determinant
print(np.linalg.det(A))

# Inverse
print(np.linalg.inv(A))
```

```python
# Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)
print(eigenvalues)
print(eigenvectors)
```

```python
# Solving linear equations (Ax = B)
C = np.array([5, 11])
solution = np.linalg.solve(A, C)
print(solution)
```

```python
arr = np.array([3, 1, 4, 1, 5, 9, 2, 6])
# Sorting an array
sorted_arr = np.sort(arr)
print(sorted_arr)
# Unique elements
unique_vals = np.unique(arr)
print(unique_vals)
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
```

```python
# Stacking arrays vertically and horizontally
vertical_stack = np.vstack((A, B))
horizontal_stack = np.hstack((A, B))
print(vertical_stack)
print(horizontal_stack)
```

```python
# Splitting arrays
split_arr = np.split(np.array([1, 2, 3, 4, 5, 6]), 3)
print(split_arr)
```

```python
arr = np.array([10, 20, 30])
# Shallow copy (view)
view_arr = arr.view()
view_arr[0] = 100
print(arr) # Changes reflect in original array
```

```python
# Deep copy
copy_arr = arr.copy()
copy_arr[0] = 200
print(arr) # Original array remains unchanged
```
